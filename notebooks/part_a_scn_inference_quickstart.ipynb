```python
# Download the slim SCN weight from the v1.0 GitHub Release
import os, urllib.request
URL = "https://github.com/rohanbalu05/fetal-head-biometry/releases/download/v1.0/part_a_scn.pth"
DST = "part_a_landmark/weights/slim/part_a_scn.pth"
os.makedirs(os.path.dirname(DST), exist_ok=True)
urllib.request.urlretrieve(URL, DST)
size = os.path.getsize(DST)
print(f"downloaded {DST}: {size:,} bytes")
assert os.path.exists(DST) and size > 1_000_000, "SCN weight download failed"
```

```python
# Run label-free inference on the shipped demo image (CC BY 4.0 HC18 frame)
!python part_a_landmark/test_part_a.py \
    --model scn \
    --weights part_a_landmark/weights/slim/part_a_scn.pth \
    --images assets/demo \
    --out-dir runs/quickstart
```

```python
# Display the predicted landmarks overlaid on the demo image
import matplotlib.pyplot as plt
import matplotlib.image as mpimg

overlay = "runs/quickstart/sample_hc18_overlay.png"
img = mpimg.imread(overlay)
plt.figure(figsize=(8, 6))
plt.imshow(img)
plt.axis("off")
plt.title("SCN predicted BPD/OFD landmarks — sample_hc18.png")
plt.show()
```
